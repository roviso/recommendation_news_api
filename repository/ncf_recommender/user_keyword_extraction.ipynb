```python
def get_user_keywords(user_id):
    user_df = user_group.get_group(user_id)
#     user_df = user_df.drop_duplicates(subset=['url'])
    keyword_array = np.array([keywords for keywords in user_df.keywords]).flatten()
#     Counter = Counter(all_arrays)
#     most_occur = Counter.most_common(20)
    user_keywords = [pre.keyword_enocder.get(keywords) for keywords,_ in Counter(keyword_array).most_common(20)]
    return user_keywords
```
