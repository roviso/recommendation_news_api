```python
pre.tfidfVectors = pd.DataFrame(X.T.toarray(), index=vectorizer.get_feature_names())
pre.tfidfVectors
```

```python
pre.tfidfVectors.shape[1]
```

```python
def get_similar_articles(q, pre):
  print("query:", q)
  print("Articles with high cosine similarity are: ")
  q = [q]
  q_vec = vectorizer.transform(q).toarray().reshape(pre.tfidfVectors.shape[0],)
  sim = {}
  for i in range(pre.tfidfVectors.shape[1]):
    sim[i] = np.dot(pre.tfidfVectors.loc[:, i].values, q_vec) / np.linalg.norm(pre.tfidfVectors.loc[:, i]) * np.linalg.norm(q_vec)
  
  sim_sorted = sorted(sim.items(), key=lambda x: x[1], reverse=True)
  
  for k, v in sim_sorted:
    if v != 0.0:
      print("Article Similarity:", v)
      print(docs[k], pre.article_df.url.iloc[k])
      print()
```

```python
def get_user_recommendation(user_keywords):
    keywords=" ".join(word for word in user_keywords)
    return get_similar_articles(keywords,pre)
```
